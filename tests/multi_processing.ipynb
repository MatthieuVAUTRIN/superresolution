```python
import os
import time
import concurrent.futures
import multiprocessing
```

```python
multiprocessing.cpu_count()
```

```python
def worker_function(return_arg):
    print(f"Worker {os.getpid()} is running\n")
    time.sleep(3)
    print(f"Worker {os.getpid()} has finished\n")
    return return_arg
```

```python
futures = []
with concurrent.futures.ProcessPoolExecutor(max_workers=10) as executor:
    for i in range(15):
        print(f"Submitting task {i}")
        futures.append(executor.submit(worker_function, i))
```

```python
results = []
num_success = 0
num_failed = 0
for future in concurrent.futures.as_completed(futures):
    try:
        return_val = future.result() # Get result (batch_idx, status)
        results.append(return_val)
        
    except Exception as exc:
        # This catches errors during process execution or result retrieval
        print(f'A batch generated an exception during execution/retrieval: {exc}')

print(f"Results: {results}")
```
