```python
import numpy as np
from sys import getsizeof
```

# create data

```python
z_arr = np.zeros((2000, 2000), dtype=np.float32)
z_arr[:, 0] = 1.0
np.save('large_array.npy', z_arr)
```

```python
print(z_arr[:, 0])
print(z_arr[0, :].shape)

print(z_arr[:, z_arr.shape[1]-1])
print(z_arr[:, z_arr.shape[1]-1].shape)
```

```python
getsizeof(z_arr) * 1e-6
```

# load

```python
load_arr = np.load('large_array.npy')
```

```python
print(type(load_arr[0][0]))
print(load_arr.shape)
```

```python
print(load_arr[:, 0])
print(load_arr[0, :].shape)

print(load_arr[:, load_arr.shape[1]-1])
print(load_arr[:, load_arr.shape[1]-1].shape)
```

# memmap

```python
# memmap_arr = np.memmap('large_array.npy', mode='r', dtype=np.float32, shape=(2000, 2000))
memmap_arr = np.lib.format.open_memmap('large_array.npy', mode='r+', dtype=np.float32, shape=(2000, 2000))
```

```python
print(type(memmap_arr[0][0]))
print(memmap_arr.shape)
```

```python
print(memmap_arr[:, 0])
print(memmap_arr[0, :].shape)

print(memmap_arr[:, memmap_arr.shape[1]-1])
print(memmap_arr[:, memmap_arr.shape[1]-1].shape)
```

```python
memmap_arr[0, 0] = 2.0
```

```python
getsizeof(memmap_arr) * 1e-6
```
